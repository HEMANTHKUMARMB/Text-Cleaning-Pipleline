```python
import re
import contractions
import nltk
import pandas as pd

from nltk.corpus import stopwords

nltk.download('stopwords')
```

```python
review = """I'm really HAPPY about this product!!! It's amazing & I don't
think I've ever used something this good. @product #awesome"""

print("Original Review:")
print(review)
```

```python
stop_words = set(stopwords.words('english'))

def clean_text(text):

    # Step 1: Count words before preprocessing
    words_before = len(text.split())

    # Step 2: Expand contractions
    # Example:
    # don't -> do not
    # can't -> cannot
    # I'm -> I am
    # it's -> it is
    text = contractions.fix(text)

    # Step 3: Case Folding
    text = text.lower()

    # Step 4: Remove special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Step 5: Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Step 6: Remove stopwords
    words = text.split()

    cleaned_words = [
        word for word in words
        if word not in stop_words
    ]

    cleaned_text = ' '.join(cleaned_words)

    # Step 7: Count words after preprocessing
    words_after = len(cleaned_text.split())

    return cleaned_text, words_before, words_after
```

```python
cleaned_review, before_count, after_count = clean_text(review)

print("Original Text:")
print(review)
```

```python
print("\nCleaned Text:")
print(cleaned_review)
```

```python
print("\nNumber of words before preprocessing:", before_count)
print("Number of words after preprocessing:", after_count)
```

```python
result = pd.DataFrame({
    "Original Text": [review],
    "Cleaned Text": [cleaned_review],
    "Words Before": [before_count],
    "Words After": [after_count]
})

result
```

```python
def clean_social_media_text(text):

    # 1. Expand contractions
    # can't -> cannot
    # wasn't -> was not
    # I'm -> I am
    text = contractions.fix(text)

    # 2. Case folding
    text = text.lower()

    # 3. Remove mentions
    # Example: @user
    text = re.sub(r'@\w+', '', text)

    # 4. Remove hashtags
    # Example: #Amazing
    text = re.sub(r'#\w+', '', text)

    # 5. Remove special characters, emojis and punctuation
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # 6. Remove unnecessary extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 7. Remove stopwords
    words = text.split()

    cleaned_words = [
        word for word in words
        if word not in stop_words
    ]

    cleaned_text = ' '.join(cleaned_words)

    return cleaned_text
```

```python
def clean_news_text(text):

    # 1. Handle contractions
    # can't -> cannot
    # won't -> will not
    # isn't -> is not
    text = contractions.fix(text)

    # 2. Case folding
    text = text.lower()

    # 3. Remove special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Remove unnecessary spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 4. Remove stopwords
    words = text.split()

    cleaned_words = [
        word for word in words
        if word not in stop_words
    ]

    cleaned_text = ' '.join(cleaned_words)

    return cleaned_text
```

```python
all_original_text = " ".join(df["article"]).lower()

# Remove special characters
all_original_text = re.sub(
    r'[^a-zA-Z\s]',
    ' ',
    all_original_text
)

words = all_original_text.split()

stopword_list = [
    word for word in words
    if word in stop_words
]

stopword_frequency = Counter(stopword_list)

top_stopwords = stopword_frequency.most_common(10)

print("Most Frequent Stopwords:")

for word, count in top_stopwords:
    print(word, ":", count)
```

```python
words_removed = total_words_before - total_words_after

reduction_percentage = (
    words_removed / total_words_before
) * 100

print("Words Before :", total_words_before)
print("Words After  :", total_words_after)
print("Words Removed:", words_removed)

print(
    "Text Size Reduction:",
    round(reduction_percentage, 2),
    "%"
)
```

```python
1.How many words were present before preprocessing?

The total number of words present in the original/raw text before performing any text-cleaning operations.

Formula:Words Before=Total number of words in original text

Python:words_before = len(original_text.split())

Example: If the original text contains 200 words:

Words Before=200
```

```python
2.How many words remained after preprocessing?

The total number of words remaining after removing stopwords, special characters, punctuation, and other unnecessary text.
Formula:Words After=Total number of words in cleaned text
Python:words_after = len(cleaned_text.split())
Example: If 120 words remain:
Words After
```

```python
3.Which 10 words occur most frequently?
The 10 words that occur the highest number of times in the cleaned text are called the top 10 most frequent words.

Formula:Frequency(w)=Number of times word w occurs
```

```python
4.Which stopwords appeared most frequently?
Stopwords are commonly occurring words that usually carry little meaning for text analysis.

Examples:the, is, a, an, and, of, to, in, was, this
The most frequent stopwords are the stopwords that appeared the greatest number of times in the original text.
Formula:Stopword Frequency(s)=Number of occurrences of stopword s
```

```python
5.How much did the text size reduce after preprocessing?

Text-size reduction tells us how much text was removed during preprocessing.

Formula for words removed
Words Removed=Words Before−Words After

Percentage reduction formula ⭐
Reduction %=Words Before/Words Before−Words After×100

Example

Suppose:Words Before=200, Words After=120

Step 1 — Find words removed:200−120=80

Step 2 — Find percentage: 200/80 ×100 =0.4×100 40%

Therefore, 80 words were removed and the text size was reduced by 40%.
```
